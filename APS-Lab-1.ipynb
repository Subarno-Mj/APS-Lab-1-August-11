**Problem 1- Coin toss for 100 times**

```python
#  to simulate random outputs
import random


def coin_tosses(num_tosses):

    # Define the two possible outcomes of a coin toss
    outcomes = ["Head", "Tail"]

    # Tossing the coin 100 times randomely using a list comprehension
    # random.choice() picks either 'Head' or 'Tail' with equal probability
    results = [random.choice(outcomes) for _ in range(num_tosses)]

    # Count the total number of occurrences for 'Head' and 'Tail'
    heads_count = results.count("Head")
    tails_count = results.count("Tail")

    # Calculate the experimental probability of getting a Head
    exp_prob_head = heads_count / num_tosses

    # Calculate the experimental probability of getting a Tail
    exp_prob_tail = tails_count / num_tosses

    # Print the detailed breakdown of the simulation results
    print("Coin Toss Results for 100 trails ---")
    print(f"Total Tosses   : {num_tosses}")
    print(f"Number of Heads   : {heads_count}")
    print(f"Number of Tails   : {tails_count}")

    # Note every time the experimental Probability changes
    print(f"Experimental Probability of Head = {exp_prob_head:.2f} or ({exp_prob_head * 100:.2f}%)")  #.2f for 2 decimal places
    print(f"Experimental Probability of Tail = {exp_prob_tail:.2f} or ({exp_prob_tail * 100:.2f}%)")

# To CALL & Execute the function
coin_tosses(100)
```

**Problem 3 - Rolling a fair Dice for 6,000 times**

```python
# To generate random die rolls
import random


def dice_rolls(total_rolls):

    # Define the outcomes of a standard die
    outcomes = [1, 2, 3, 4, 5, 6]

    # Initialize a dictionary to store the count of each outcome, starting with initial value of 0
    """" Create a Python dictionary comprehension.
    It creates a new dictionary where each item in the outcomes list or set becomes a key,
     and every key gets an initial value of 0
     """
    counts = {face: 0 for face in outcomes}

    # Roll the dice to Generate a random integer between 1 to 6
    for _ in range(total_rolls):
        roll_result = random.randint(1, 6)
        counts[roll_result] += 1   # Increment the count of the outcome which is the key

    # Loop through outcomes 1 to 6 to calculate and display experimental probabilities
    for face in outcomes:
        # experimental probability = (occurrences of outcome) / (total rolls)
        exp_prob = counts[face] / total_rolls

        # Calculate percentage representation
        percentage = exp_prob * 100


         # Print the results

        print("-" * 60)
        print(f"{'Outcome':<10} | {'Count':<10} | {'Experimental Prob':<20} | {'Percentage':<10}")

        print(f"Face {face:<5} | {counts[face]:<10} | {exp_prob:<20.4f} | {percentage:.2f}%")


# Execute for 6000 rolls
dice_rolls(6000)
```

**Problem 4 - Addition Rule for Disjoint Events**

```python
import random

def verify_disjoint_events(num_rolls):

    # rolling a 6-sided die for 10K times
    rolls = [random.randint(1, 6) for _ in range(num_rolls)]

    # Initialize counters for our events
    count_A = 0          # Rolling a 1
    count_B = 0          # Rolling a 6
    count_A_and_B = 0    # Rolling a 1 AND a 6 on the same roll (Intersection)
    count_A_or_B = 0     # Rolling a 1 OR a 6 (Union)

    # count our events , joint , disjoint events :-
    for roll in rolls:
        if roll == 1:
            count_A += 1
        if roll == 6:
            count_B += 1

        # count for intersection
        if roll == 1 and roll == 6:
            count_A_and_B += 1

        # count for Union
        if roll == 1 or roll == 6:
            count_A_or_B += 1

    # Calculate Experimental Probabilities
    p_A = count_A / num_rolls
    p_B = count_B / num_rolls
    p_A_and_B = count_A_and_B / num_rolls
    p_A_or_B = count_A_or_B / num_rolls

    # Print the counts and calculated probabilities
    print(f"Event Counts for ({num_rolls} dice rolls) ---")
    print(f"Event A (Rolling a 1) Count : {count_A}")
    print(f"Event B (Rolling a 6) Count : {count_B}")
    print(f"Event A AND B (Intersection): {count_A_and_B}")
    print(f"Event A OR B (Union)        : {count_A_or_B}\n")

    print("Experimental Probabilities ---")
    print(f"P(A)  = {p_A:.4f}")
    print(f"P(B)  = {p_B:.4f}")
    print(f"P(A AND B) = {p_A_and_B:.4f}")
    print(f"P(A OR B)  = {p_A_or_B:.4f}\n")

    # Check if the events are disjoint
    print("Checking Joint vs. Disjoint ---")
    if p_A_and_B == 0:
        print("Result: P(A AND B) is 0. The events cannot happen simultaneously.")
        print("Conclusion: Event A and Event B are DISJOINT (Mutually Exclusive).\n")
    else:
        print("Result: P(A AND B) > 0. The events are JOINT.\n")

    # Verify the Addition Rule for Disjoint Events: P(A U B) = P(A) + P(B)
    print("Verifying P(A U B) = P(A) + P(B) ---")
    p_A_plus_B = p_A + p_B
    print(f"P(A U B) calculated from experiment : {p_A_or_B:.4f}")
    print(f"P(A) + P(B) mathematically added    : {p_A_plus_B:.4f}")

    # Use a small tolerance adjustment for floating point math comparisons to verify
    if round(p_A_or_B, 6) == round(p_A_plus_B, 6):
        print("Verification: SUCCESS! P(A U B) exactly equals P(A) + P(B).")
    else:
        print("Verification: FAILED.")

# Call the function
verify_disjoint_events(10000)
```

# Problem 5 - Addition Rule for Overlapping Events

```python
# To verify the General Addition Rule for a single die roll


def verify_general_addition_rule():
    # Define the sample space
    sample_space = {1, 2, 3, 4, 5, 6}

    total_outcomes = len(sample_space)

    # Define Event A (getting an even number) and Event B (getting 4, 5, or 6)
    event_A = {2, 4, 6}
    event_B = {4, 5, 6}

    # Function to Find the Intersection (A ∩ B):  present in both A AND B
    intersection_A_B = event_A.intersection(event_B)

    # Find the Union (A ∪ B): present in A OR B
    union_A_B = event_A.union(event_B)

    # Calculate probabilities
    p_A = len(event_A) / total_outcomes
    p_B = len(event_B) / total_outcomes

    p_intersection = len(intersection_A_B) / total_outcomes
    p_union_actual = len(union_A_B) / total_outcomes

    # Calculate P(A ∪ B) =  P(A) + P(B) - P(A ∩ B)
    p_union_formula = p_A + p_B - p_intersection

    # Display Event details
    print("Event Details ---")
    print(f"Sample Space (S)      : {sample_space}")
    print(f"Event A {event_A}")
    print(f"Event B {event_B}")
    print(f"Intersection (A ∩ B)  : {intersection_A_B}")
    print(f"Union (A ∪ B)         : {union_A_B}\n")

    # Display individual probabilities
    print("Calculated Probabilities ---")
    print(f"P(A)  : {len(event_A)}/{total_outcomes} = {p_A:.4f}")
    print(f"P(B)  : {len(event_B)}/{total_outcomes} = {p_B:.4f}")
    print(f"P(A ∩ B) : {len(intersection_A_B)}/{total_outcomes} = {p_intersection:.4f}")
    print(f"P(A ∪ B) [Direct Set]  : {len(union_A_B)}/{total_outcomes} = {p_union_actual:.4f}\n")

    # Verify the formula
    print("Verification of Formula: P(A ∪ B) = P(A) + P(B) - P(A ∩ B) ")
    print(f"Right Hand Side: P(A) + P(B) - P(A ∩ B) = {p_A:.4f} + {p_B:.4f} - {p_intersection:.4f} = {p_union_formula:.4f}")
    print(f"Left Hand Side : P(A ∪ B) = {p_union_actual:.4f}")

    # Use round() to avoid floating-point comparison issues
    if round(p_union_actual, 6) == round(p_union_formula, 6):
        print("\nVerification SUCCESS: Both sides are equal!")
    else:
        print("\nVerification FAILED.")


# call the function
verify_general_addition_rule()
```

# Problem 6 - Multiplication Rule for Independent Events

```python
import random


def verify_independent_events(num_trials):

    # Possible outcomes for the coin and die
    coin_outcomes = ["H", "T"]
    die_outcomes = [1, 2, 3, 4, 5, 6]

    # Initialize counters for Event A, Event B, and their Intersection (A ∩ B)
    count_A = 0  # Count of getting a Head
    count_B = 0  # Count of getting an even number (2, 4, 6)
    count_A_and_B = 0   # Count of getting BOTH a Head AND an even number

    # Perform the trails for 10k times each
    for _ in range(num_trials):
        # Toss the coin and roll the die independently
        coin_result = random.choice(coin_outcomes)
        die_result = random.choice(die_outcomes)

        # Check Event A:
        is_event_A = coin_result == "H"

        # Check Event B:
        is_event_B = die_result in [2, 4, 6]

        # Update event counters based on current trial results
        if is_event_A:
            count_A += 1

        if is_event_B:
            count_B += 1

        # Check Intersection (A ∩ B): Both conditions met in the same trial
        if is_event_A and is_event_B:
            count_A_and_B += 1

    # Calculate Experimental Probabilities
    p_A = count_A / num_trials
    p_B = count_B / num_trials
    p_A_and_B = count_A_and_B / num_trials

    # Calculate the product P(A) * P(B)
    p_A_times_p_B = p_A * p_B

    # Display event counts and experimental probabilities
    print(f"Results ({num_trials} Trials) ---")
    print(f"Event A Count (Heads) : {count_A}")
    print(f"Event B Count (Even numbers) : {count_B}")
    print(f"Intersection Count (Heads AND Even): {count_A_and_B}\n")

    print("Experimental Probabilities ---")
    print(f"P(A)  = {p_A:.4f}")
    print(f"P(B)  = {p_B:.4f}")
    print(f"P(A ∩ B) [Joint Prob] = {p_A_and_B:.4f}")
    print(f"P(A) × P(B) [Calculated Product]= {p_A_times_p_B:.4f}\n")

    # Verify if P(A ∩ B) = P(A) * P(B)
    print("Independence events Verification ---")
    if abs(p_A_and_B - p_A_times_p_B) < 0.01:
        print("Verification: SUCCESS!")
        print(f"P(A ∩ B) [{p_A_and_B:.4f}] is practically equal to P(A) × P(B) [{p_A_times_p_B:.4f}].")
        print("Conclusion: Events A and B are INDEPENDENT (Coin toss does not affect die roll).")
    else:
        print("Verification: FAILED")


# call function
verify_independent_events(10000)
```
